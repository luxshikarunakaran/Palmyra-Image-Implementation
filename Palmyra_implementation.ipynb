```python
# %% [markdown]
# # TP-ORD — Thermal-Privileged, Rank-Consistent Distillation for Palmyra Palm Health Grading
#
# **Research idea (new):** thermal imaging carries the physiological signal of palm decline
# (reduced transpiration → less evaporative cooling → warmer canopy; Red Palm Weevil larval
# fermentation → internal trunk heating), but thermal cameras are expensive and their readings
# shift with time of day / ambient temperature. RGB cameras are cheap but miss early stress.
#
# TP-ORD therefore:
# 1. **SITN — Session-Invariant Thermal Normalisation** + **affine-invariant thermal physiology
#    descriptors** (hot-spot contrast, heat asymmetry, skew/kurtosis, hot-blob compactness,
#    crown-vs-trunk gradient). They are unchanged under T' = a·T + b, so a different time slot or
#    ambient temperature does not change them.
# 2. **Biophysical RGB tokens** — vegetation indices (ExG, ExGR, VARI, GLI, NGRDI) and
#    chlorosis / necrosis / desiccation colour fractions.
# 3. **Multimodal ordinal fusion transformer** over frozen *foundation* (DINOv2 ViT-S/14,
#    self-supervised) + *efficient* (MobileNetV4) embeddings and the biophysical tokens, with
#    **modality dropout**, **CORN rank-consistent ordinal loss** (Non_infected < Infected <
#    Badly_damaged < Dead), and **supervised contrastive** regularisation → the *teacher*.
# 4. **Cross-fitted, rank-consistent privileged distillation**: out-of-fold teacher predictions
#    (no train-set overconfidence) are distilled into an **RGB-only student** through both a
#    softened-softmax KL and an *ordinal conditional-probability* KD term → deployable without a
#    thermal camera.
# 5. **Trustworthy evaluation**: tree/near-duplicate **group-aware CV** (no leakage), temperature
#    calibration, ECE, **split-conformal prediction sets** with coverage guarantees, QWK for
#    ordinal agreement, group bootstrap CIs, McNemar tests, ablations, and a **biological
#    hypothesis analysis** (Kruskal–Wallis + Spearman trend + BH-FDR at tree level).
#
# **CPU-fast design:** backbones are frozen and run once (embeddings cached to Drive); all learning
# happens in tiny heads that train in seconds. Re-running after a runtime reset reuses the cache.

# %% [markdown]
# ## 0. Install

# %%
# !pip -q install "timm>=1.0.9" opencv-python-headless scikit-learn scipy pandas matplotlib seaborn tqdm joblib

# %% [markdown]
# ## 1. Configuration

# %%
import os, re, sys, json, time, math, random, hashlib, warnings
from pathlib import Path
import numpy as np
import pandas as pd
import cv2
from PIL import Image
import torch
import torch.nn as nn
import torch.nn.functional as F
import timm
from tqdm.auto import tqdm
from joblib import Parallel, delayed
from scipy import stats
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.inspection import permutation_importance
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score, cohen_kappa_score,
                             roc_auc_score, log_loss, confusion_matrix, classification_report)
from sklearn.manifold import TSNE
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")
SMOKE = os.environ.get("PALM_SMOKE") == "1"          # internal quick test switch
IN_COLAB = "google.colab" in sys.modules
if IN_COLAB:
    from google.colab import drive
    drive.mount("/content/drive")

DATA_ROOT = Path(os.environ.get("PALM_ROOT",
                 "/content/drive/MyDrive/Palmyra dataset/Processed Data/Processed Data"))
OUT_DIR = Path(os.environ.get("PALM_OUT", "/content/drive/MyDrive/Palmyra dataset/TPORD_outputs"))
CACHE_DIR, FIG_DIR, DEPLOY_DIR = OUT_DIR / "cache", OUT_DIR / "figures", OUT_DIR / "deploy"
for d in (CACHE_DIR, FIG_DIR, DEPLOY_DIR):
    d.mkdir(parents=True, exist_ok=True)

CLASSES = ["Non_infected", "Infected", "Badly_damaged", "Dead"]   # ordinal severity order
K = len(CLASSES)
MODALITY_DIRS = {"rgb": "RGB", "thermal": "THERMAL"}

BACKBONES = {                                    # frozen feature extractors
    "dinov2_s": "vit_small_patch14_dinov2.lvd142m",       # self-supervised foundation ViT
    "mnv4_s":   "mobilenetv4_conv_small.e2400_r224_in1k",  # efficient 2024 CNN
}
IMG_SIZE = 224
BATCH = 32
USE_TTA = False            # True = average with horizontal flip (2x extraction time)
N_FOLDS = 5
SEED = 42
GROUP_REGEX = None         # e.g. r"(T\d+)" if file names contain a tree ID; None = auto near-duplicate grouping
DHASH_THRESH = 6           # Hamming distance (of 64 bits) for near-duplicate grouping
RUN_SITN_ABLATION = True   # extra thermal pass with the raw pseudo-colour images
RUN_ABLATIONS = True
EPOCHS, PATIENCE = 150, 25
CONF_ALPHA = 0.10          # conformal: target 90% coverage
MAX_PER_CLASS = None       # e.g. 50 for a quick dry run
NUM_WORKERS = min(4, os.cpu_count() or 1)

if SMOKE:
    EPOCHS, PATIENCE, N_FOLDS = 8, 4, 3
    NUM_WORKERS = 0

torch.set_num_threads(os.cpu_count() or 1)

def seed_all(s=SEED):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
seed_all()
print("torch", torch.__version__, "| timm", timm.__version__, "| threads", torch.get_num_threads())

# %% [markdown]
# ## 2. Index images and pair RGB ↔ thermal

# %%
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}

def index_modality(mod):
    rows = []
    for ci, c in enumerate(CLASSES):
        d = DATA_ROOT / MODALITY_DIRS[mod] / c
        assert d.exists(), f"Missing folder: {d}"
        files = sorted(p for p in d.rglob("*") if p.suffix.lower() in IMG_EXT)
        if MAX_PER_CLASS:
            files = files[:MAX_PER_CLASS]
        rows += [dict(label=ci, path=str(p), stem=p.stem) for p in files]
    return pd.DataFrame(rows)

rgb_df, th_df = index_modality("rgb"), index_modality("thermal")
print("RGB per class:\n", rgb_df.label.map(dict(enumerate(CLASSES))).value_counts())
print("THERMAL per class:\n", th_df.label.map(dict(enumerate(CLASSES))).value_counts())

MOD_TOKENS = r"(rgb|thermal|therm|lwir|ir|visible|vis|colour|color)"
def key_exact(s):  return s.lower()
def key_clean(s):
    s = re.sub(MOD_TOKENS, "", s.lower())
    return re.sub(r"[^a-z0-9]", "", s) or None
def key_digits(s):
    d = re.findall(r"\d+", s)
    return "_".join(str(int(x)) for x in d) if d else None

def pair_modalities(rgb, th):
    best = (None, pd.DataFrame())
    for name, fn in [("exact", key_exact), ("clean", key_clean), ("digits", key_digits)]:
        a = rgb.assign(key=rgb.stem.map(fn)).dropna(subset=["key"]).drop_duplicates(["label", "key"], keep=False)
        b = th.assign(key=th.stem.map(fn)).dropna(subset=["key"]).drop_duplicates(["label", "key"], keep=False)
        m = a.merge(b, on=["label", "key"], suffixes=("_rgb", "_th"))
        if len(m) > len(best[1]):
            best = (name, m)
    name, m = best
    frac = len(m) / max(1, min(len(rgb), len(th)))
    if frac >= 0.3:
        return name, m[["label", "path_rgb", "path_th", "stem_rgb", "stem_th"]]
    same = rgb.label.value_counts().sort_index().equals(th.label.value_counts().sort_index())
    if same:
        parts = []
        for c in range(K):
            a = rgb[rgb.label == c].sort_values("stem").reset_index(drop=True)
            b = th[th.label == c].sort_values("stem").reset_index(drop=True)
            parts.append(pd.DataFrame(dict(label=c, path_rgb=a.path, path_th=b.path,
                                           stem_rgb=a.stem, stem_th=b.stem)))
        print("WARNING: no filename match; pairing by sorted order inside each class (counts are equal). "
              "Check a few pairs visually below.")
        return "sorted-order", pd.concat(parts, ignore_index=True)
    return None, None

pair_mode, pairs = pair_modalities(rgb_df, th_df)
if pairs is not None:
    df = pairs.reset_index(drop=True)
    PAIRED = True
else:   # unpaired: union of both sets, missing modality is masked inside the fusion model
    PAIRED = False
    df = pd.concat([
        pd.DataFrame(dict(label=rgb_df.label, path_rgb=rgb_df.path, path_th=None, stem_rgb=rgb_df.stem, stem_th=None)),
        pd.DataFrame(dict(label=th_df.label, path_rgb=None, path_th=th_df.path, stem_rgb=None, stem_th=th_df.stem)),
    ], ignore_index=True)
    print("WARNING: RGB and thermal could not be paired -> running in UNPAIRED mode "
          "(fusion uses masked tokens; distillation gain will be limited).")

y = df.label.values.astype(int)
N = len(df)
has_rgb = df.path_rgb.notna().values
has_th = df.path_th.notna().values
print(f"pair mode = {pair_mode} | samples = {N} | paired = {PAIRED} | rgb={has_rgb.sum()} thermal={has_th.sum()}")
df.head()

# %%
# Visual sanity check of pairs (one per class)
if PAIRED:
    fig, ax = plt.subplots(2, K, figsize=(3.2 * K, 6.2))
    for c in range(K):
        r = df[df.label == c].iloc[0]
        ax[0, c].imshow(cv2.cvtColor(cv2.imread(r.path_rgb), cv2.COLOR_BGR2RGB)); ax[0, c].set_title(f"{CLASSES[c]}\nRGB", fontsize=9)
        ax[1, c].imshow(cv2.cvtColor(cv2.imread(r.path_th), cv2.COLOR_BGR2RGB)); ax[1, c].set_title("THERMAL", fontsize=9)
        ax[0, c].axis("off"); ax[1, c].axis("off")
    plt.tight_layout(); plt.savefig(FIG_DIR / "pairs_check.png", dpi=150); plt.show()

# %% [markdown]
# ## 3. Leakage-aware grouping (same tree / near-duplicate images stay in one fold)
# The Processed Data may contain several views of the same tree (trunk/top/crown, several time
# slots) or augmented copies. Random splits would leak. We group by a tree ID regex if given,
# otherwise by perceptual-hash near-duplicate clustering (flip-aware), using union–find.

# %%
def dhash_variants(path, size=8):
    img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
    img = cv2.resize(img, (size + 1, size + 1), interpolation=cv2.INTER_AREA)
    out = []
    for v in (img, img[:, ::-1], img[::-1, :]):          # original, h-flip, v-flip
        out.append((v[:size, 1:] > v[:size, :-1]).flatten())
    return np.stack(out).astype(np.float32)               # 3 x 64

def near_duplicate_groups(paths, thresh=DHASH_THRESH):
    H = np.stack(Parallel(n_jobs=-1)(delayed(dhash_variants)(p) for p in tqdm(paths, desc="dHash")))
    n = len(paths); parent = np.arange(n)
    def find(i):
        while parent[i] != i:
            parent[i] = parent[parent[i]]; i = parent[i]
        return i
    base = H[:, 0]                                       # n x 64
    for v in range(3):
        T = H[:, v]
        for s in range(0, n, 1024):
            A = base[s:s + 1024]
            agree = A @ T.T + (1 - A) @ (1 - T).T         # matching bits
            ii, jj = np.where((64 - agree) <= thresh)
            for i, j in zip(ii + s, jj):
                if i != j:
                    ri, rj = find(i), find(j)
                    if ri != rj: parent[ri] = rj
    return np.array([find(i) for i in range(n)])

hash_paths = np.where(has_rgb, df.path_rgb.fillna(""), df.path_th.fillna(""))
if GROUP_REGEX:
    stems = df.stem_rgb.fillna(df.stem_th)
    groups = stems.map(lambda s: (re.search(GROUP_REGEX, s) or [s])[0]).astype("category").cat.codes.values
else:
    groups = near_duplicate_groups(list(hash_paths))
groups = pd.Series(groups).astype("category").cat.codes.values
print(f"{N} images -> {len(np.unique(groups))} groups (largest group = {np.bincount(groups).max()} images)")

# %% [markdown]
# ## 4. Biological / biophysical features
# **RGB (plant physiology):** chlorophyll loss and canopy decline lower greenness indices
# (ExG, ExGR, VARI, GLI, NGRDI); chlorosis → yellow fraction; necrosis → brown fraction;
# desiccated dead fronds → grey/low-saturation fraction.
#
# **Thermal (thermo-physiology), affine-invariant:** stressed palms close stomata and transpire less,
# so canopy temperature rises relative to surroundings; RPW larval activity creates internal hot
# spots; dead palms lose thermoregulation and look thermally flat or heated uniformly by the sun.
# Every descriptor below is a ratio of temperature *differences* (or a standardised moment), so it is
# invariant to T' = a·T + b — i.e., to camera auto-gain and to the session's ambient offset.

# %%
def _stats(x, name):
    return {f"{name}_mean": float(x.mean()), f"{name}_std": float(x.std()),
            f"{name}_p10": float(np.percentile(x, 10)), f"{name}_p90": float(np.percentile(x, 90))}

def rgb_bio(path):
    bgr = cv2.resize(cv2.imread(path), (256, 256), interpolation=cv2.INTER_AREA)
    rgb = bgr[..., ::-1].astype(np.float32) / 255.0
    R, G, B = rgb[..., 0], rgb[..., 1], rgb[..., 2]
    S = R + G + B + 1e-6
    r, g, b = R / S, G / S, B / S
    exg = 2 * g - r - b
    exgr = exg - (1.4 * r - g)
    vari = np.clip((G - R) / (G + R - B + 1e-6), -1, 1)
    gli = (2 * G - R - B) / (2 * G + R + B + 1e-6)
    ngrdi = (G - R) / (G + R + 1e-6)
    hsv = cv2.cvtColor(bgr, cv2.COLOR_BGR2HSV)
    h, s, v = hsv[..., 0], hsv[..., 1] / 255.0, hsv[..., 2] / 255.0
    colored = (s > 0.15) & (v > 0.15)
    f = {}
    for n, a in [("ExG", exg), ("ExGR", exgr), ("VARI", vari), ("GLI", gli), ("NGRDI", ngrdi)]:
        f.update(_stats(a, "rgb_" + n))
    f["rgb_green_frac"] = float(((h >= 30) & (h <= 85) & colored).mean())
    f["rgb_yellow_frac"] = float(((h >= 20) & (h < 30) & colored).mean())   # chlorosis
    f["rgb_brown_frac"] = float(((h >= 5) & (h < 20) & colored).mean())     # necrosis
    f["rgb_grey_frac"] = float((s < 0.12).mean())                           # desiccation
    f["rgb_sat_mean"], f["rgb_val_mean"] = float(s.mean()), float(v.mean())
    gray = cv2.cvtColor(bgr, cv2.COLOR_BGR2GRAY).astype(np.float32)
    f["rgb_texture"] = float(cv2.Laplacian(gray, cv2.CV_32F).var() / (gray.var() + 1e-6))
    return f

def load_thermal_scalar(path):
    im = cv2.imread(path, cv2.IMREAD_UNCHANGED)
    if im.ndim == 3:
        im = cv2.cvtColor(im[..., :3], cv2.COLOR_BGR2GRAY)   # luminance as relative temperature proxy
    return im.astype(np.float32)

def thermal_bio(path):
    t = cv2.resize(load_thermal_scalar(path), (256, 256), interpolation=cv2.INTER_AREA)
    med = np.median(t)
    q5, q25, q75, q95 = np.percentile(t, [5, 25, 75, 95])
    iqr, rng = q75 - q25 + 1e-6, q95 - q5 + 1e-6
    z = (t - med) / iqr
    tn = np.clip((t - q5) / rng, 0, 1).astype(np.float32)
    f = {
        "th_hot_contrast": (q95 - med) / iqr,
        "th_cold_contrast": (med - q5) / iqr,
        "th_heat_asymmetry": ((q95 - med) - (med - q5)) / rng,
        "th_rel_heat_index": (t.mean() - q5) / rng,
        "th_skew": float(stats.skew(t.ravel())),
        "th_kurtosis": float(stats.kurtosis(t.ravel())),
        "th_hot_frac": float((z > 1.5).mean()),
        "th_cold_frac": float((z < -1.5).mean()),
    }
    hist = np.histogram(tn, bins=32, range=(0, 1))[0] / tn.size
    f["th_entropy"] = float(-(hist[hist > 0] * np.log2(hist[hist > 0])).sum())
    gx, gy = cv2.Sobel(tn, cv2.CV_32F, 1, 0), cv2.Sobel(tn, cv2.CV_32F, 0, 1)
    f["th_grad_mean"] = float(np.sqrt(gx ** 2 + gy ** 2).mean())
    f["th_lap_var"] = float(cv2.Laplacian(tn, cv2.CV_32F).var())
    hh, ww = tn.shape
    c = tn[hh // 4:3 * hh // 4, ww // 4:3 * ww // 4]
    f["th_centre_minus_periph"] = float(c.mean() - (tn.sum() - c.sum()) / (tn.size - c.size))
    f["th_top_minus_bottom"] = float(tn[:hh // 2].mean() - tn[hh // 2:].mean())   # crown vs trunk
    hot = (z > 1.5).astype(np.uint8)
    n, _, st, _ = cv2.connectedComponentsWithStats(hot, 8)
    f["th_largest_hot_blob"] = float(st[1:, cv2.CC_STAT_AREA].max() / hot.size) if n > 1 else 0.0
    f["th_n_hot_blobs"] = float(n - 1)
    return {k: float(v) for k, v in f.items()}

def bio_matrix(paths, fn, tag):
    f = CACHE_DIR / f"bio_{tag}_{hashlib.md5(''.join(paths).encode()).hexdigest()[:10]}.pkl"
    if f.exists():
        return pd.read_pickle(f)
    out = pd.DataFrame(Parallel(n_jobs=-1)(delayed(fn)(p) for p in tqdm(paths, desc=f"bio {tag}")))
    out.to_pickle(f)
    return out

def expand(mask, arr_valid, ncols):
    full = np.full((N, ncols), np.nan, dtype=np.float32)
    full[mask] = arr_valid
    return full

rgb_bio_df = bio_matrix(list(df.path_rgb[has_rgb]), rgb_bio, "rgb")
th_bio_df = bio_matrix(list(df.path_th[has_th]), thermal_bio, "th")
RGB_BIO_COLS, TH_BIO_COLS = list(rgb_bio_df.columns), list(th_bio_df.columns)
B_rgb_nan = expand(has_rgb, rgb_bio_df.values, len(RGB_BIO_COLS))
B_th_nan = expand(has_th, th_bio_df.values, len(TH_BIO_COLS))
print("bio features:", len(RGB_BIO_COLS), "RGB +", len(TH_BIO_COLS), "thermal")

# %% [markdown]
# ## 5. Frozen deep embeddings (DINOv2 + MobileNetV4), cached
# Thermal images go through **SITN**: per-image robust percentile normalisation of the temperature
# proxy (1st–99th percentile → [0,1]), replicated to 3 channels. This removes the session's offset and
# the colour-palette auto-scaling. The ablation (`th_emb_raw`) feeds the original pseudo-colour file.

# %%
class ImgDS(torch.utils.data.Dataset):
    def __init__(self, paths, tf, mode):
        self.paths, self.tf, self.mode = list(paths), tf, mode
    def __len__(self):
        return len(self.paths)
    def __getitem__(self, i):
        p = self.paths[i]
        if self.mode == "thermal_sitn":
            t = load_thermal_scalar(p)
            lo, hi = np.percentile(t, [1, 99])
            t = np.clip((t - lo) / (hi - lo + 1e-6), 0, 1)
            img = (np.repeat(t[..., None], 3, -1) * 255).astype(np.uint8)
        else:   # rgb or raw thermal file
            img = cv2.cvtColor(cv2.imread(p, cv2.IMREAD_COLOR), cv2.COLOR_BGR2RGB)
        return self.tf(Image.fromarray(img))

def build_backbone(model_id):
    kw = dict(img_size=IMG_SIZE) if "vit" in model_id else {}
    m = timm.create_model(model_id, pretrained=not SMOKE, num_classes=0, **kw).eval()
    cfg = timm.data.resolve_data_config({}, model=m)
    cfg.update(input_size=(3, IMG_SIZE, IMG_SIZE), crop_pct=1.0)
    return m, timm.data.create_transform(**cfg)

@torch.inference_mode()
def extract(bb_name, model_id, paths, mode):
    h = hashlib.md5(("".join(paths) + mode + str(USE_TTA) + str(SMOKE)).encode()).hexdigest()[:10]
    f = CACHE_DIR / f"emb_{bb_name}_{mode}_{h}.npy"
    if f.exists():
        return np.load(f)
    model, tf = build_backbone(model_id)
    dl = torch.utils.data.DataLoader(ImgDS(paths, tf, mode), batch_size=BATCH, num_workers=NUM_WORKERS)
    out = []
    t0 = time.time()
    for x in tqdm(dl, desc=f"{bb_name}:{mode}"):
        x = x.contiguous(memory_format=torch.channels_last)
        e = model(x)
        if USE_TTA:
            e = (e + model(torch.flip(x, dims=[3]))) / 2
        out.append(e.float().numpy())
    arr = np.concatenate(out).astype(np.float32)
    np.save(f, arr)
    print(f"  {bb_name}:{mode} {arr.shape} in {time.time() - t0:.0f}s")
    return arr

def multi_backbone(paths, mode):
    return np.concatenate([extract(n, mid, paths, mode) for n, mid in BACKBONES.items()], axis=1)

_rgb = multi_backbone(list(df.path_rgb[has_rgb]), "rgb")
_th = multi_backbone(list(df.path_th[has_th]), "thermal_sitn")
E_rgb, E_th = expand(has_rgb, _rgb, _rgb.shape[1]), expand(has_th, _th, _th.shape[1])
BLOCKS = {"rgb_emb": E_rgb, "th_emb": E_th, "rgb_bio": B_rgb_nan, "th_bio": B_th_nan}
MASKS = {"rgb_emb": has_rgb, "th_emb": has_th, "rgb_bio": has_rgb, "th_bio": has_th}
if RUN_SITN_ABLATION:
    _thr = multi_backbone(list(df.path_th[has_th]), "thermal_raw")
    BLOCKS["th_emb_raw"], MASKS["th_emb_raw"] = expand(has_th, _thr, _thr.shape[1]), has_th
print({k: v.shape for k, v in BLOCKS.items()})

# %% [markdown]
# ## 6. Models: ordinal multimodal fusion transformer, losses, training utilities

# %%
D_MODEL = 128

class FusionNet(nn.Module):
    """Tokens = one per feature block (+CLS). Missing / dropped modalities are key-padding-masked."""
    def __init__(self, dims, d=D_MODEL, heads=4, layers=2, p_in=0.2, p_mod=0.25):
        super().__init__()
        self.names = list(dims)
        self.proj = nn.ModuleDict({n: nn.Sequential(nn.LayerNorm(dims[n]), nn.Dropout(p_in),
                                                    nn.Linear(dims[n], d), nn.GELU(), nn.Linear(d, d))
                                   for n in self.names})
        self.mod_emb = nn.Parameter(torch.randn(len(self.names), d) * 0.02)
        self.cls = nn.Parameter(torch.zeros(1, 1, d))
        layer = nn.TransformerEncoderLayer(d, heads, 2 * d, dropout=0.1, batch_first=True,
                                           norm_first=True, activation="gelu")
        self.enc = nn.TransformerEncoder(layer, layers, enable_nested_tensor=False)
        self.norm = nn.LayerNorm(d)
        self.ce_head, self.corn_head = nn.Linear(d, K), nn.Linear(d, K - 1)
        self.p_mod = p_mod

    def forward(self, feats, masks):
        x = torch.stack([self.proj[n](feats[n]) + self.mod_emb[i] for i, n in enumerate(self.names)], 1)
        m = torch.stack([masks[n] for n in self.names], 1)
        if self.training and self.p_mod > 0:
            keep = m & (torch.rand(m.shape) > self.p_mod)
            empty = keep.sum(1) == 0
            keep[empty] = m[empty]
            m = keep
        B = x.shape[0]
        x = torch.cat([self.cls.expand(B, -1, -1), x], 1)
        pad = torch.cat([torch.zeros(B, 1, dtype=torch.bool), ~m], 1)
        z = self.norm(self.enc(x, src_key_padding_mask=pad)[:, 0])
        return z, self.ce_head(z), self.corn_head(z)

def corn_loss(logits, y):
    """CORN (Shi et al.): conditional binary tasks P(y>j | y>j-1)."""
    tot, cnt = 0.0, 0
    for j in range(K - 1):
        sel = y >= j
        if sel.any():
            tot = tot + F.binary_cross_entropy_with_logits(logits[sel, j], (y[sel] > j).float(), reduction="sum")
            cnt += int(sel.sum())
    return tot / max(cnt, 1)

def corn_probs(logits):
    cum = torch.cumprod(torch.sigmoid(logits), 1)             # P(y>j)
    p = torch.cat([1 - cum[:, :1], cum[:, :-1] - cum[:, 1:], cum[:, -1:]], 1).clamp_min(1e-7)
    return p / p.sum(1, keepdim=True)

def supcon_loss(z, y, t=0.1):
    z = F.normalize(z, dim=1)
    sim = z @ z.T / t
    eye = torch.eye(len(y), dtype=torch.bool)
    sim = sim.masked_fill(eye, -1e9)
    pos = (y[:, None] == y[None, :]) & ~eye
    logp = sim - torch.logsumexp(sim, 1, keepdim=True)
    npos = pos.sum(1)
    ok = npos > 0
    return -((logp * pos).sum(1)[ok] / npos[ok]).mean() if ok.any() else z.sum() * 0

def ordinal_kd(corn_logits, p_teacher):
    """Rank-consistent KD: match student P(y>j|y>j-1) to the teacher's conditional probabilities."""
    surv = 1 - torch.cumsum(p_teacher, 1)[:, :-1]             # P(y>j), j=0..K-2
    prev = torch.cat([torch.ones_like(surv[:, :1]), surv[:, :-1]], 1)
    cond = (surv / prev.clamp_min(1e-6)).clamp(1e-4, 1 - 1e-4)
    w = prev.detach()                                         # weight by reachability of the task
    bce = F.binary_cross_entropy_with_logits(corn_logits, cond, reduction="none")
    return (bce * w).sum() / w.sum()

def fit_scalers(idx, names):
    sc = {}
    for n in names:
        X = BLOCKS[n][idx][MASKS[n][idx]]
        sc[n] = (np.nanmean(X, 0), np.nanstd(X, 0) + 1e-6)
    return sc

def make_tensors(idx, names, sc):
    feats = {n: torch.tensor(np.clip(np.nan_to_num((BLOCKS[n][idx] - sc[n][0]) / sc[n][1]), -8, 8),
                             dtype=torch.float32) for n in names}
    masks = {n: torch.tensor(MASKS[n][idx]) for n in names}
    return feats, masks

def head_probs(ce, corn):
    return 0.5 * F.softmax(ce, 1) + 0.5 * corn_probs(corn)

@torch.no_grad()
def predict(model, names, idx, sc, drop=None):
    feats, masks = make_tensors(idx, names, sc)
    if drop:
        masks[drop] = torch.zeros_like(masks[drop])
    model.eval()
    z, ce, corn = model(feats, masks)
    return head_probs(ce, corn).numpy(), z.numpy()

def train_head(names, tr_idx, va_idx=None, fixed_epochs=None, lam_corn=1.0, lam_sup=0.1, p_mod=0.25,
               kd_probs=None, alpha=1.0, beta=1.0, T_kd=3.0, seed=SEED, bs=64, lr=1e-3):
    seed_all(seed)
    sc = fit_scalers(tr_idx, names)
    dims = {n: BLOCKS[n].shape[1] for n in names}
    model = FusionNet(dims, p_mod=p_mod)
    Xtr, Mtr = make_tensors(tr_idx, names, sc)
    ytr = torch.tensor(y[tr_idx])
    Ptr = torch.tensor(kd_probs[tr_idx], dtype=torch.float32) if kd_probs is not None else None
    cnt = np.bincount(y[tr_idx], minlength=K).astype(float)
    cw = torch.tensor(len(tr_idx) / (K * np.maximum(cnt, 1)), dtype=torch.float32)
    epochs = fixed_epochs or EPOCHS
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=0.05)
    sched = torch.optim.lr_scheduler.OneCycleLR(opt, max_lr=lr, total_steps=epochs * math.ceil(len(tr_idx) / bs),
                                                pct_start=0.1)
    best, best_state, best_ep, bad = -1, None, epochs, 0
    for ep in range(1, epochs + 1):
        model.train()
        perm = torch.randperm(len(tr_idx))
        for s in range(0, len(perm), bs):
            b = perm[s:s + bs]
            if len(b) < 2:
                continue
            fb = {n: Xtr[n][b] + 0.05 * torch.randn_like(Xtr[n][b]) for n in names}   # feature-noise aug
            mb = {n: Mtr[n][b] for n in names}
            z, ce, corn = model(fb, mb)
            loss = F.cross_entropy(ce, ytr[b], weight=cw, label_smoothing=0.05)
            if lam_corn: loss = loss + lam_corn * corn_loss(corn, ytr[b])
            if lam_sup:  loss = loss + lam_sup * supcon_loss(z, ytr[b])
            if Ptr is not None:
                q = Ptr[b] ** (1 / T_kd); q = q / q.sum(1, keepdim=True)
                loss = loss + alpha * (T_kd ** 2) * F.kl_div(F.log_softmax(ce / T_kd, 1), q, reduction="batchmean")
                if beta: loss = loss + beta * ordinal_kd(corn, Ptr[b])
            opt.zero_grad(); loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            opt.step(); sched.step()
        if va_idx is not None and fixed_epochs is None:
            p, _ = predict(model, names, va_idx, sc)
            score = f1_score(y[va_idx], p.argmax(1), average="macro") + 0.5 * cohen_kappa_score(
                y[va_idx], p.argmax(1), weights="quadratic")
            if score > best + 1e-4:
                best, best_ep, bad = score, ep, 0
                best_state = {k: v.clone() for k, v in model.state_dict().items()}
            else:
                bad += 1
                if bad >= PATIENCE:
                    break
    if best_state is not None:
        model.load_state_dict(best_state)
    return model, sc, best_ep

def fit_temperature(p, yy):
    """Temperature scaling on probabilities: p_T ∝ p^(1/T), T chosen by NLL."""
    grid = np.exp(np.linspace(np.log(0.05), np.log(10), 80))
    nll = [log_loss(yy, apply_T(p, T), labels=range(K)) for T in grid]
    return float(grid[int(np.argmin(nll))])

def apply_T(p, T):
    q = np.clip(p, 1e-8, 1) ** (1 / T)
    return q / q.sum(1, keepdims=True)

def conformal_qhat(p_cal, y_cal, alpha=CONF_ALPHA):
    s = 1 - p_cal[np.arange(len(y_cal)), y_cal]
    n = len(s)
    return float(np.quantile(s, min(1, np.ceil((n + 1) * (1 - alpha)) / n), method="higher"))

def conformal_sets(p, qhat):
    sets = p >= 1 - qhat
    sets[np.arange(len(p)), p.argmax(1)] = True        # never empty
    return sets

# %% [markdown]
# ## 7. Metrics

# %%
def ece_score(yy, p, bins=15):
    conf, acc = p.max(1), p.argmax(1) == yy
    e = 0.0
    for lo, hi in zip(np.linspace(0, 1, bins + 1)[:-1], np.linspace(0, 1, bins + 1)[1:]):
        m = (conf > lo) & (conf <= hi)
        if m.any():
            e += m.mean() * abs(acc[m].mean() - conf[m].mean())
    return float(e)

def all_metrics(yy, p):
    pred = p.argmax(1)
    out = dict(acc=accuracy_score(yy, pred), bal_acc=balanced_accuracy_score(yy, pred),
               f1_macro=f1_score(yy, pred, average="macro"), f1_weighted=f1_score(yy, pred, average="weighted"),
               qwk=cohen_kappa_score(yy, pred, weights="quadratic"), mae_ord=float(np.abs(yy - pred).mean()),
               ece=ece_score(yy, p), brier=float(((p - np.eye(K)[yy]) ** 2).sum(1).mean()),
               logloss=log_loss(yy, np.clip(p, 1e-8, 1), labels=range(K)))
    try:
        out["auc_macro"] = roc_auc_score(yy, p, multi_class="ovr", average="macro")
    except ValueError:
        out["auc_macro"] = np.nan
    return out

def group_bootstrap_ci(yy, p, g, fn, B=1000, seed=SEED):
    rng = np.random.default_rng(seed)
    ug = np.unique(g)
    by = {k: np.where(g == k)[0] for k in ug}
    vals = []
    for _ in range(B):
        ii = np.concatenate([by[k] for k in rng.choice(ug, len(ug), replace=True)])
        vals.append(fn(yy[ii], p[ii]))
    return np.percentile(vals, [2.5, 97.5])

def mcnemar(yy, pa, pb):
    a, b = pa.argmax(1) == yy, pb.argmax(1) == yy
    n01, n10 = int((a & ~b).sum()), int((~a & b).sum())
    pval = stats.binomtest(n01, n01 + n10, 0.5).pvalue if n01 + n10 > 0 else 1.0
    return n01, n10, pval

# %% [markdown]
# ## 8. Leakage-aware cross-validation of all models
# Inside each outer fold: an inner group split gives a validation set used for early stopping,
# temperature calibration and conformal calibration. The teacher's soft labels for the student are
# produced by **inner 3-fold cross-fitting** so the student never distils an over-fitted teacher.

# %%
FULL = ["rgb_emb", "th_emb", "rgb_bio", "th_bio"]
RGB_ONLY = ["rgb_emb", "rgb_bio"]

def rows_with(idx, names):
    return idx[np.any(np.stack([MASKS[n][idx] for n in names], 1), 1)]

def inner_split(idx, n_splits=6, seed=SEED):
    sg = StratifiedGroupKFold(n_splits=n_splits, shuffle=True, random_state=seed)
    a, b = next(sg.split(idx, y[idx], groups[idx]))
    return idx[a], idx[b]

def crossfit_teacher(names, idx, epochs, n_inner=3, seed=SEED):
    P = np.full((N, K), 1.0 / K, dtype=np.float32)
    sg = StratifiedGroupKFold(n_splits=n_inner, shuffle=True, random_state=seed)
    for a, b in sg.split(idx, y[idx], groups[idx]):
        m, sc, _ = train_head(names, idx[a], fixed_epochs=epochs, seed=seed)
        P[idx[b]] = predict(m, names, idx[b], sc)[0]
    return P

def run_neural(tag, names, tr, va, te, store, conf_store, kd=None, **kw):
    tr_, va_, te_ = rows_with(tr, names), rows_with(va, names), rows_with(te, names)
    m, sc, ep = train_head(names, tr_, va_, kd_probs=kd, **kw)
    T = fit_temperature(predict(m, names, va_, sc)[0], y[va_])
    p_va, p_te = apply_T(predict(m, names, va_, sc)[0], T), apply_T(predict(m, names, te_, sc)[0], T)
    store[tag][te_] = p_te
    q = conformal_qhat(p_va, y[va_])
    sets = conformal_sets(p_te, q)
    conf_store.append(dict(model=tag, fold=FOLD, qhat=q, T=T, best_epoch=ep,
                           coverage=float(sets[np.arange(len(te_)), y[te_]].mean()), avg_set=float(sets.sum(1).mean())))
    return m, sc, ep, T, q

def lr_model():
    return make_pipeline(StandardScaler(), LogisticRegression(C=0.5, max_iter=3000, class_weight="balanced"))

MODELS = ["LR-RGB", "LR-Thermal", "LR-RGB+Thermal", "HGB-Bio", "TP-ORD Teacher (RGB+T)",
          "RGB Transformer (no KD)", "TP-ORD Student (RGB-only, KD)"]
if RUN_ABLATIONS:
    MODELS += ["Abl: Teacher -bio tokens", "Abl: Teacher CE only", "Abl: Teacher no modality-dropout",
               "Abl: Student logit-KD only"]
    if RUN_SITN_ABLATION:
        MODELS += ["Abl: Teacher raw thermal (no SITN)"]
OOF = {m: np.full((N, K), np.nan, dtype=np.float32) for m in MODELS}
fold_of = np.full(N, -1)
CONF, MOD_ABL, Z_TEACHER, BEST = [], [], np.full((N, D_MODEL), np.nan, np.float32), []

outer = StratifiedGroupKFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)
t_start = time.time()
for FOLD, (tr_all, te) in enumerate(outer.split(np.zeros(N), y, groups)):
    t0 = time.time()
    fold_of[te] = FOLD
    tr, va = inner_split(tr_all, seed=SEED + FOLD)

    # ---- classical baselines (trained on tr+va) ----
    for tag, cols in [("LR-RGB", ["rgb_emb"]), ("LR-Thermal", ["th_emb"]), ("LR-RGB+Thermal", ["rgb_emb", "th_emb"])]:
        a, b = rows_with(tr_all, cols), rows_with(te, cols)
        a = a[np.all(np.stack([MASKS[c][a] for c in cols], 1), 1)]
        b = b[np.all(np.stack([MASKS[c][b] for c in cols], 1), 1)]
        if len(a) and len(b):
            clf = lr_model().fit(np.hstack([BLOCKS[c][a] for c in cols]), y[a])
            OOF[tag][b] = clf.predict_proba(np.hstack([BLOCKS[c][b] for c in cols]))
    Xbio = np.hstack([B_rgb_nan, B_th_nan])               # HGB handles NaN natively
    hgb = HistGradientBoostingClassifier(max_iter=300, learning_rate=0.05, class_weight="balanced",
                                         random_state=SEED).fit(Xbio[tr_all], y[tr_all])
    OOF["HGB-Bio"][te] = hgb.predict_proba(Xbio[te])

    # ---- TP-ORD teacher ----
    tm, tsc, t_ep, _, _ = run_neural("TP-ORD Teacher (RGB+T)", FULL, tr, va, te, OOF, CONF)
    te_full = rows_with(te, FULL)
    Z_TEACHER[te_full] = predict(tm, FULL, te_full, tsc)[1]
    base_f1 = f1_score(y[te_full], predict(tm, FULL, te_full, tsc)[0].argmax(1), average="macro")
    for blk in FULL:                                      # test-time modality ablation
        p = predict(tm, FULL, te_full, tsc, drop=blk)[0]
        MOD_ABL.append(dict(fold=FOLD, dropped=blk,
                            f1_drop=base_f1 - f1_score(y[te_full], p.argmax(1), average="macro")))

    # ---- cross-fitted privileged soft labels (only training rows) ----
    tr_va = np.concatenate([tr, va])
    P_kd = crossfit_teacher(FULL, rows_with(tr_va, FULL), epochs=t_ep, seed=SEED + FOLD)

    # ---- RGB-only models ----
    run_neural("RGB Transformer (no KD)", RGB_ONLY, tr, va, te, OOF, CONF)
    _, _, s_ep, _, _ = run_neural("TP-ORD Student (RGB-only, KD)", RGB_ONLY, tr, va, te, OOF, CONF, kd=P_kd)
    BEST.append(dict(fold=FOLD, teacher_ep=t_ep, student_ep=s_ep))

    # ---- ablations ----
    if RUN_ABLATIONS:
        run_neural("Abl: Teacher -bio tokens", ["rgb_emb", "th_emb"], tr, va, te, OOF, CONF)
        run_neural("Abl: Teacher CE only", FULL, tr, va, te, OOF, CONF, lam_corn=0, lam_sup=0)
        run_neural("Abl: Teacher no modality-dropout", FULL, tr, va, te, OOF, CONF, p_mod=0.0)
        run_neural("Abl: Student logit-KD only", RGB_ONLY, tr, va, te, OOF, CONF, kd=P_kd, beta=0)
        if RUN_SITN_ABLATION:
            run_neural("Abl: Teacher raw thermal (no SITN)", ["rgb_emb", "th_emb_raw", "rgb_bio", "th_bio"],
                       tr, va, te, OOF, CONF)
    print(f"fold {FOLD + 1}/{N_FOLDS} done in {time.time() - t0:.0f}s  "
          f"(teacher F1={base_f1:.3f})")
print(f"CV total {time.time() - t_start:.0f}s")

# %% [markdown]
# ## 9. Results tables (per-fold mean ± sd, pooled OOF with 95% group-bootstrap CI)

# %%
rows, fold_rows = [], []
for m in MODELS:
    ok = ~np.isnan(OOF[m][:, 0])
    if not ok.any():
        continue
    yy, pp, gg = y[ok], OOF[m][ok], groups[ok]
    met = all_metrics(yy, pp)
    lo, hi = group_bootstrap_ci(yy, pp, gg, lambda a, b: f1_score(a, b.argmax(1), average="macro"))
    qlo, qhi = group_bootstrap_ci(yy, pp, gg, lambda a, b: cohen_kappa_score(a, b.argmax(1), weights="quadratic"))
    rows.append(dict(model=m, n=int(ok.sum()), **{k: round(v, 4) for k, v in met.items()},
                     f1_CI=f"[{lo:.3f}, {hi:.3f}]", qwk_CI=f"[{qlo:.3f}, {qhi:.3f}]"))
    for f in range(N_FOLDS):
        sel = ok & (fold_of == f)
        if sel.any():
            fold_rows.append(dict(model=m, fold=f, **all_metrics(y[sel], OOF[m][sel])))
res = pd.DataFrame(rows).set_index("model")
fold_df = pd.DataFrame(fold_rows)
summ = fold_df.groupby("model")[["acc", "f1_macro", "qwk", "bal_acc", "ece"]].agg(["mean", "std"]).round(4)
conf_df = pd.DataFrame(CONF)
conf_summary = conf_df.groupby("model")[["coverage", "avg_set", "T"]].mean().round(3)
res.to_csv(OUT_DIR / "results_pooled.csv"); fold_df.to_csv(OUT_DIR / "results_per_fold.csv", index=False)
conf_df.to_csv(OUT_DIR / "conformal_per_fold.csv", index=False)
pd.set_option("display.width", 250)
print("=== Pooled out-of-fold metrics ===")
display(res) if "display" in globals() else print(res)
print("=== Per-fold mean / sd ===")
display(summ) if "display" in globals() else print(summ)
print(f"=== Split-conformal (target coverage {1 - CONF_ALPHA:.0%}) ===")
display(conf_summary) if "display" in globals() else print(conf_summary)

# %%
# Statistical comparisons on identical rows
def compare(a, b):
    ok = ~np.isnan(OOF[a][:, 0]) & ~np.isnan(OOF[b][:, 0])
    n01, n10, pv = mcnemar(y[ok], OOF[a][ok], OOF[b][ok])
    rng = np.random.default_rng(SEED)
    ug = np.unique(groups[ok]); by = {k: np.where(groups[ok] == k)[0] for k in ug}
    ya, pa, pb = y[ok], OOF[a][ok], OOF[b][ok]
    ds = []
    for _ in range(1000):
        ii = np.concatenate([by[k] for k in rng.choice(ug, len(ug))])
        ds.append(f1_score(ya[ii], pa[ii].argmax(1), average="macro") - f1_score(ya[ii], pb[ii].argmax(1), average="macro"))
    return dict(A=a, B=b, A_only_correct=n01, B_only_correct=n10, mcnemar_p=round(pv, 5),
                dF1_mean=round(float(np.mean(ds)), 4), dF1_CI=f"[{np.percentile(ds, 2.5):.3f}, {np.percentile(ds, 97.5):.3f}]")

pairs_to_test = [("TP-ORD Student (RGB-only, KD)", "RGB Transformer (no KD)"),
                 ("TP-ORD Student (RGB-only, KD)", "LR-RGB"),
                 ("TP-ORD Teacher (RGB+T)", "LR-RGB+Thermal"),
                 ("TP-ORD Teacher (RGB+T)", "TP-ORD Student (RGB-only, KD)")]
if RUN_ABLATIONS and RUN_SITN_ABLATION:
    pairs_to_test.append(("TP-ORD Teacher (RGB+T)", "Abl: Teacher raw thermal (no SITN)"))
cmp_df = pd.DataFrame([compare(a, b) for a, b in pairs_to_test])
cmp_df.to_csv(OUT_DIR / "statistical_tests.csv", index=False)
print(cmp_df.to_string(index=False))
for m in ["TP-ORD Teacher (RGB+T)", "TP-ORD Student (RGB-only, KD)"]:
    ok = ~np.isnan(OOF[m][:, 0])
    print(f"\n--- {m} ---\n", classification_report(y[ok], OOF[m][ok].argmax(1), target_names=CLASSES, digits=4))

# %% [markdown]
# ## 10. Figures

# %%
def save(name):
    plt.tight_layout(); plt.savefig(FIG_DIR / name, dpi=200, bbox_inches="tight"); plt.show()

# Confusion matrices
show = ["LR-RGB", "TP-ORD Teacher (RGB+T)", "TP-ORD Student (RGB-only, KD)"]
fig, ax = plt.subplots(1, 3, figsize=(17, 4.8))
for a, m in zip(ax, show):
    ok = ~np.isnan(OOF[m][:, 0])
    cm = confusion_matrix(y[ok], OOF[m][ok].argmax(1), labels=range(K), normalize="true")
    sns.heatmap(cm, annot=True, fmt=".2f", cmap="Greens", xticklabels=CLASSES, yticklabels=CLASSES, ax=a, cbar=False)
    a.set_title(m, fontsize=10); a.set_xlabel("Predicted"); a.set_ylabel("True")
save("confusion_matrices.png")

# Model comparison with CIs
plot = res.reset_index()
lo_hi = np.array([[float(v) for v in s.strip("[]").split(",")] for s in plot.f1_CI])
plt.figure(figsize=(9, 0.45 * len(plot) + 1))
plt.barh(plot.model, plot.f1_macro, xerr=[plot.f1_macro - lo_hi[:, 0], lo_hi[:, 1] - plot.f1_macro],
         color=["#2e7d32" if "TP-ORD" in m else ("#9e9e9e" if "Abl" in m else "#64b5f6") for m in plot.model])
plt.xlabel("Macro-F1 (pooled OOF, 95% group-bootstrap CI)"); plt.gca().invert_yaxis()
save("model_comparison_f1.png")

# Reliability diagram: student
m = "TP-ORD Student (RGB-only, KD)"
ok = ~np.isnan(OOF[m][:, 0])
conf_, acc_ = OOF[m][ok].max(1), OOF[m][ok].argmax(1) == y[ok]
edges = np.linspace(0, 1, 11); mids, accs = [], []
for lo, hi in zip(edges[:-1], edges[1:]):
    s = (conf_ > lo) & (conf_ <= hi)
    if s.any(): mids.append(conf_[s].mean()); accs.append(acc_[s].mean())
plt.figure(figsize=(4.5, 4.5)); plt.plot([0, 1], [0, 1], "k--", lw=1); plt.plot(mids, accs, "o-", c="#2e7d32")
plt.xlabel("Confidence"); plt.ylabel("Accuracy"); plt.title(f"Reliability — student (ECE={ece_score(y[ok], OOF[m][ok]):.3f})")
save("reliability_student.png")

# Test-time modality importance in the teacher
mab = pd.DataFrame(MOD_ABL)
plt.figure(figsize=(6, 3.5)); sns.barplot(data=mab, x="dropped", y="f1_drop", color="#ef6c00", errorbar="sd")
plt.ylabel("Macro-F1 drop when removed"); plt.title("Teacher: modality contribution")
save("modality_importance.png")

# t-SNE of teacher representations (out-of-fold)
ok = ~np.isnan(Z_TEACHER[:, 0])
idx = np.where(ok)[0]
if len(idx) > 3000:
    idx = np.random.default_rng(SEED).choice(idx, 3000, replace=False)
emb2 = TSNE(2, perplexity=min(30, max(5, len(idx) // 10)), random_state=SEED, init="pca").fit_transform(Z_TEACHER[idx])
plt.figure(figsize=(6, 5))
for c in range(K):
    s = y[idx] == c
    plt.scatter(emb2[s, 0], emb2[s, 1], s=8, label=CLASSES[c], alpha=0.7)
plt.legend(markerscale=2); plt.title("t-SNE of TP-ORD teacher embedding (OOF)"); plt.xticks([]); plt.yticks([])
save("tsne_teacher.png")

# %% [markdown]
# ## 11. Biological analysis — do thermal/colour physiology signatures track severity?
# Tested at **tree/group level** (mean per group) to avoid pseudo-replication. Kruskal–Wallis across
# the four classes (effect size ε²), Spearman ρ with the ordinal severity, Benjamini–Hochberg FDR.

# %%
bio_all = pd.DataFrame(np.hstack([B_rgb_nan, B_th_nan]), columns=RGB_BIO_COLS + TH_BIO_COLS)
bio_all["y"], bio_all["g"] = y, groups
gl = bio_all.groupby("g").agg({**{c: "mean" for c in RGB_BIO_COLS + TH_BIO_COLS},
                               "y": lambda s: s.mode().iloc[0]})
recs = []
for c in RGB_BIO_COLS + TH_BIO_COLS:
    d = gl[[c, "y"]].dropna()
    samples = [d[c][d.y == k].values for k in range(K) if (d.y == k).sum() > 1]
    if len(samples) < 2 or d[c].nunique() < 3:
        continue
    H, p = stats.kruskal(*samples)
    eps2 = (H - len(samples) + 1) / (len(d) - len(samples))
    rho, prho = stats.spearmanr(d[c], d.y)
    recs.append(dict(feature=c, modality="thermal" if c.startswith("th_") else "rgb", KW_H=H, KW_p=p,
                     eps2=eps2, spearman_rho=rho, spearman_p=prho,
                     **{f"median_{CLASSES[k]}": d[c][d.y == k].median() for k in range(K)}))
bio_stats = pd.DataFrame(recs)
if len(bio_stats):
    bio_stats["KW_q_BH"] = stats.false_discovery_control(bio_stats.KW_p)
    bio_stats = bio_stats.sort_values("eps2", ascending=False)
    bio_stats.to_csv(OUT_DIR / "biological_feature_tests.csv", index=False)
    print(bio_stats.head(15).round(4).to_string(index=False))

    top = bio_stats.feature.head(8).tolist()
    long = bio_all[top + ["y"]].melt(id_vars="y").dropna()
    long["class"] = long.y.map(dict(enumerate(CLASSES)))
    g = sns.catplot(data=long, x="class", y="value", col="variable", col_wrap=4, kind="box",
                    sharey=False, height=3, aspect=1.1, order=CLASSES, palette="YlOrBr", showfliers=False)
    g.set_xticklabels(rotation=30); g.set_titles("{col_name}")
    plt.savefig(FIG_DIR / "bio_feature_boxplots.png", dpi=200, bbox_inches="tight"); plt.show()

# Permutation importance of biophysical features (HGB on all data, group-held-out)
tr_i, te_i = inner_split(np.arange(N), n_splits=5)
hgb = HistGradientBoostingClassifier(max_iter=300, learning_rate=0.05, class_weight="balanced",
                                     random_state=SEED).fit(Xbio[tr_i], y[tr_i])
pi = permutation_importance(hgb, Xbio[te_i], y[te_i], scoring="f1_macro", n_repeats=10, random_state=SEED, n_jobs=-1)
imp = pd.Series(pi.importances_mean, index=RGB_BIO_COLS + TH_BIO_COLS).sort_values(ascending=False).head(15)
plt.figure(figsize=(6, 5)); imp[::-1].plot.barh(color=["#d84315" if i.startswith("th_") else "#43a047" for i in imp.index[::-1]])
plt.xlabel("Permutation importance (Δ macro-F1)"); plt.title("Biophysical feature importance")
save("bio_permutation_importance.png")

# %% [markdown]
# ## 12. Train the final deployable RGB-only student on all data and save it

# %%
bests = pd.DataFrame(BEST)
t_ep, s_ep = int(bests.teacher_ep.median()), int(bests.student_ep.median())
all_idx = np.arange(N)
P_kd_all = crossfit_teacher(FULL, rows_with(all_idx, FULL), epochs=t_ep, n_inner=5)
rgb_rows = rows_with(all_idx, RGB_ONLY)
final_student, final_sc, _ = train_head(RGB_ONLY, rgb_rows, fixed_epochs=s_ep, kd_probs=P_kd_all)
st_conf = conf_df[conf_df.model == "TP-ORD Student (RGB-only, KD)"]
T_final, q_final = float(st_conf["T"].median()), float(st_conf.qhat.median())
torch.save(final_student.state_dict(), DEPLOY_DIR / "tpord_student.pt")
np.savez(DEPLOY_DIR / "tpord_student_scalers.npz",
         **{f"{n}_mu": final_sc[n][0] for n in RGB_ONLY}, **{f"{n}_sd": final_sc[n][1] for n in RGB_ONLY})
meta = dict(classes=CLASSES, backbones=BACKBONES, img_size=IMG_SIZE, names=RGB_ONLY,
            dims={n: int(BLOCKS[n].shape[1]) for n in RGB_ONLY}, rgb_bio_cols=RGB_BIO_COLS,
            temperature=T_final, conformal_qhat=q_final, conformal_alpha=CONF_ALPHA,
            teacher_epochs=t_ep, student_epochs=s_ep, pair_mode=pair_mode, n_samples=int(N),
            n_groups=int(len(np.unique(groups))))
json.dump(meta, open(DEPLOY_DIR / "tpord_meta.json", "w"), indent=2)
print("saved to", DEPLOY_DIR, "\n", json.dumps(meta, indent=1)[:600])

# %% [markdown]
# ## 13. Inference on a new RGB image (no thermal camera needed)

# %%
_BB_CACHE = {}
@torch.inference_mode()
def predict_rgb_image(path):
    embs = []
    for n, mid in BACKBONES.items():
        if n not in _BB_CACHE:
            _BB_CACHE[n] = build_backbone(mid)
        m, tf = _BB_CACHE[n]
        x = tf(Image.open(path).convert("RGB")).unsqueeze(0)
        e = m(x)
        if USE_TTA:
            e = (e + m(torch.flip(x, dims=[3]))) / 2
        embs.append(e.numpy())
    raw = {"rgb_emb": np.concatenate(embs, 1),
           "rgb_bio": np.array([[rgb_bio(path)[c] for c in RGB_BIO_COLS]], dtype=np.float32)}
    feats = {n: torch.tensor(np.clip((raw[n] - final_sc[n][0]) / final_sc[n][1], -8, 8), dtype=torch.float32)
             for n in RGB_ONLY}
    masks = {n: torch.ones(1, dtype=torch.bool) for n in RGB_ONLY}
    final_student.eval()
    _, ce, corn = final_student(feats, masks)
    p = apply_T(head_probs(ce, corn).numpy(), T_final)[0]
    pset = [CLASSES[k] for k in np.where(conformal_sets(p[None], q_final)[0])[0]]
    severity = float((p * np.arange(K)).sum())         # expected ordinal severity 0..3
    return dict(prediction=CLASSES[int(p.argmax())], probabilities=dict(zip(CLASSES, p.round(4).tolist())),
                expected_severity=round(severity, 3), conformal_set_90=pset)

ex = df.path_rgb.dropna().iloc[0]
print(ex, "\n", predict_rgb_image(ex))

# %%
summary = dict(pair_mode=pair_mode, n=int(N), groups=int(len(np.unique(groups))),
               pooled=res[["acc", "f1_macro", "qwk", "bal_acc", "ece", "auc_macro", "f1_CI"]].to_dict("index"),
               conformal=conf_summary.to_dict("index"))
json.dump(summary, open(OUT_DIR / "summary.json", "w"), indent=2, default=float)
print("All outputs in:", OUT_DIR)
```
